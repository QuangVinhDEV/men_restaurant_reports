```python
import pandas as pd


# Đọc file (vì bạn dùng ; làm separator)
df_revenue_per_date = pd.read_csv('../data/bronze_layer/revenue_per_date_table.csv', sep=';')

# Đổi tên cột sang tiếng Anh
df_revenue_per_date = df_revenue_per_date.rename(columns={
    'Ngày': 'date',
    'Tiền thuế GTGT': 'vat',
    'Thực thu': 'revenue',
    'Số khách hàng': 'covers'
})

df_revenue_per_date['date'] = pd.to_datetime(df_revenue_per_date['date'], format='%d/%m/%Y')
print(df_revenue_per_date.tail())
```

```python
# create the column with name the TC
tc_mapping = df_combined.groupby('date')['bill_id'].nunique()

df_revenue_per_date['TC'] = df_revenue_per_date['date'].map(tc_mapping).fillna(0).astype(int)

print(df_revenue_per_date.tail())
```

```python
# 1. Loại bỏ dấu chấm, sau đó mới chuyển sang int
# Chúng ta dùng .str.replace('.', '') để xóa dấu chấm
df_revenue_per_date['revenue'] = df_revenue_per_date['revenue'].astype(str).str.replace('.', '', regex=False).astype(int)
df_revenue_per_date['vat'] = df_revenue_per_date['vat'].astype(str).str.replace('.', '', regex=False).astype(int)

# 2. Bây giờ mới tính cột ac
df_revenue_per_date['ac'] = df_revenue_per_date['revenue'] / df_revenue_per_date['TC']

# 3. Xử lý inf/nan và làm tròn ac (đưa về int nếu muốn)
df_revenue_per_date['ac'] = df_revenue_per_date['ac'].replace([float('inf'), -float('inf')], 0).fillna(0).round(0).astype(int)

print(df_revenue_per_date.tail(20))
```
